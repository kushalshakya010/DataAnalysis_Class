```python
import pandas as pd
```

#### 1. Pandas Series

A **Series** is a one-dimensional labeled array (like a single table column with an index).

**Create a Series**

```python
s = pd.Series([10, 20, 30, 40])
s
```

**Create a Series with a custom index**

```python
marks = pd.Series([85, 90, 88], index=["Ananda", "Sita", "Ram"])
marks
```

**Access a value by position**

```python
# Get the first value by position
marks.iloc[0]
```

#### 2. Pandas DataFrame

A **DataFrame** is a two-dimensional table made of rows and columns (similar to an Excel or SQL table).

**Create a DataFrame**

```python
df = pd.DataFrame({
    "Name": ["Ananda", "Sita", "Ram"],
    "Age": [22, 21, 23],
    "Score": [85, 90, 85]
})

df
```

#### Select Columns

- A single column returns a **Series**.
- Multiple columns return a **DataFrame**.

```python
# Select one column
df["Age"]
```

```python
# Select multiple columns
df[["Name", "Score"]]
```

#### Common DataFrame Methods

Use these methods to quickly explore a DataFrame.

```python
# First 2 rows
df.head(2)
```

```python
# Last 2 rows
df.tail(2)
```

```python
# Number of rows and columns
df.shape
```

```python
# One random row
df.sample(1)
```
